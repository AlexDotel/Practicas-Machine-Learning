```python
import matplotlib.pyplot as plt
import numpy as np
fig, ax = plt.subplots()

#lista de dias de la semana
dias = ['Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes', 'Sábado', 'Domingo']
#crea un diccionario con tres ciudades y sus temperaturas
temperaturas = {
    'Madrid'    : [20, 21, 19, 22, 23, 24, 25],
    'Barcelona' : [18, 19, 20, 21, 22, 23, 24],
    'Soria'     : [15, 16, 17, 18, 19, 20, 21]
}

#Cuales son los marcadores disponibles en matplotlib
#https://matplotlib.org/stable/api/markers_api.html

ax.scatter(dias, temperaturas['Soria'], color='green', marker='*')
ax.plot(dias, temperaturas['Madrid'], color='red', marker='v', linestyle='dashed')
ax.scatter(dias, temperaturas['Barcelona'], color='blue', marker='o')
ax.set_title('Temperaturas de la semana', fontsize=20, loc='center',
             fontdict={'family': 'Consolas', 'weight': 'bold', 'size': 20, 'color': 'gray'})
ax.set_ylim([10, 25])
ax.set_xlim(0,4)

ax.set_ylabel('Temperaturas', fontsize=10, loc='center', fontdict={'family': 'Consolas', 'weight': 'bold', 'size': 15, 'color': 'gray'})
ax.set_xlabel('Días de la semana', fontsize=10, loc='center', fontdict={'family': 'Consolas', 'weight': 'bold', 'size': 15, 'color': 'gray'})

ax.set_yticks(range(5,20, 5))

plt.show()
```
